00 - Audit laterality recovery
#
# Four of the twelve targets are medial/lateral pairs, and medial is not a
# fixed side of the image -- it is whichever side faces the other leg. Right
# knees therefore have to be mirrored onto a left-knee convention before the
# model can separate those targets at all.
#
# The `Laterality` tag is Type 2C and whole vendors omit it, so the side is
# recovered from geometry where the tag is missing. This kernel measures that
# recovery against the tag where the tag does exist, before a full rebuild is
# committed to. Minutes, not hours.

```python
import sys, subprocess
from pathlib import Path

CODE = next(
    (p for pat in ("rsna-knee-code", "*/rsna-knee-code", "*/*/rsna-knee-code")
     for p in sorted(Path("/kaggle/input").glob(pat)) if (p / "src").is_dir()),
    None,
)
assert CODE is not None, "rsna-knee-code dataset is not attached to this notebook"
sys.path.insert(0, str(CODE / "src"))
sys.path.insert(0, str(CODE))
print("CODE", CODE)

# timm ships in the Kaggle image; the install is a fallback for images that lack
# it, and cannot run in the offline submission notebook.
try:
    import timm
except ImportError:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "timm"], check=False)
    import timm

import rsna_knee.config as cfg
print("RAW ", cfg.RAW, cfg.RAW.exists())
print("PREP", cfg.PREP, cfg.PREP.exists())
print("SIZE", cfg.SIZE, cfg.MOSAIC_EXT)
```
